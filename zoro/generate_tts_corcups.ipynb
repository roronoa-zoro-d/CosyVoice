# 根据讯飞识别客服的结果，进行合成

```python
import sys
import os
import openpyxl
import random

def process_text2(texts):
    kefu = []
    kehu = []
    for text in texts.strip().split("\n"):
        buff = text.split(':')
        if buff[0] == "客服":
            kefu.append(buff[1])
        elif buff[0] == "客户":
            kehu.append(buff[1])

    return kefu, kehu

excel_file = '/home/zhangjiayuan/temp/VoiceDocument0328.xlsx'
out_dir = 'tts_corcups/'
workbook = openpyxl.load_workbook(excel_file)
sheet = workbook.active

header_info = [cell.value for cell in sheet[1]]

datas = []
datas_corcups = []
for i in range(2, sheet.max_row + 1):
    xun = sheet.cell(row=i, column=8).value
    kefu, kehu = process_text2(xun)
    datas.append([kefu, kehu])
    datas_corcups.extend(kefu)
    datas_corcups.extend(kehu)

print(datas[0][0])
print(datas_corcups[0], len(datas_corcups))
out_file = f'{out_dir}/finetune_xunfei_corcups.txt'
with open(out_file, 'w', encoding='utf-8') as f:
    for text in datas_corcups:
        if len(text) > 5:
            f.write(f'{text}\n')
```
