```python
import glob
import os
import json
import random
import re

ZH_ROOT_DIR = '/data/nas/emilia/Amphion___Emilia/raw/ZH/'
json_files = glob.glob(os.path.join(ZH_ROOT_DIR, '*.jsonl'))
print(len(json_files), json_files)

pattern = r'[a-zA-Z0-9]'

spk2data = {}
for json_file in json_files:
    with open(json_file, 'r', encoding='utf-8') as file:
        for line in file:
            data = json.loads(line)
            spk = data['speaker']
            text = data['text']
            dur = data['duration']
            if len(text)>10 and len(text)<20 and  re.search(pattern, text) is None:
                spk2data.setdefault(spk, []).append(data)
    break

select_spk2datas = {}
for spk, datas in spk2data.items():
    random.shuffle(datas)
    select_spk2datas[spk] = datas[:2]
    
out_file = 'select_spk.jsonl'

with open(out_file, 'w', encoding='utf-8') as file:
    for spk, datas in select_spk2datas.items():
        for data in datas:
            file.write(json.dumps(data, ensure_ascii=False) + '\n')
```

```python
import numpy as np
print(len(select_spk2datas.keys()))
durs = []
for spk, datas in select_spk2datas.items():
    for data in datas:
        durs.append(data['duration'])
durs = np.array(durs)
print(f'total {len(durs)} max dur : {durs.max()}')
```
